```python
from sys import getsizeof
from random import choice, choices
from requests import get
from glob import glob
from collections import defaultdict
```

```python
%%time

whitelist_chars = "abcdefghijklmnopqrstuvwxyz ABCDEFGHIJKLMNOPQRSTUVWXYZ1234567890,.\"':"
i = 0

def get_source_text(link):
    global i
    text = get(link).text
    processed_text = "".join(filter(whitelist_chars.__contains__, text.lower()))
    with open(f"sources/source_{i}.txt", "w") as f:
        f.write(processed_text)
    i += 1

with open("sources.txt", "r") as f:
    sources_list = [t.strip() for t in f.readlines()]

for link in sources_list:
    get_source_text(link)
```

```python
%%time

source_token_size = 6
output_token_size = 4

analysis_dict = defaultdict(lambda: defaultdict(int))

def stream_file_to_analysis_dict(filename, chunk_size=8192):
    tail = ""

    with open(filename, "r", encoding="utf-8") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break

            chunk = "".join(filter(whitelist_chars.__contains__, chunk.lower()))
            data = tail + chunk

            if len(data) < source_token_size + output_token_size:
                tail = data
                continue

            max_i = len(data) - source_token_size - output_token_size + 1
            for i in range(max_i):
                k = data[i : i + source_token_size]
                v = data[i + source_token_size : i + source_token_size + output_token_size]
                analysis_dict[k][v] += 1

            tail = data[-(source_token_size + output_token_size - 1):]

for filename in sorted(glob("sources/source_*.txt")):
    stream_file_to_analysis_dict(filename)

analysis_dict = {k: dict(v) for k, v in analysis_dict.items()}

print(f"")
MB = 1024*1024
print(f"Dict size: {(getsizeof(analysis_dict) / MB):.2f}MB\tKey amount: {len(analysis_dict)}")
```

```python
def predict_next_token(input_token):
    input_token = input_token.lower()
    if len(input_token) < source_token_size:
        existing_keys = {key for key in analysis_dict if input_token in key}
        if len(existing_keys) == 0:
            raise Exception("Invalid input size")
        input_token = choice(list(existing_keys))
    if len(input_token) > source_token_size:
        input_token = input_token[:source_token_size]
    if input_token not in analysis_dict:
        random_key = choice(list(analysis_dict.keys()))
        return choice(list(analysis_dict[random_key].keys())) # return one token instead gibberish, and hope next itteration matches something
    frequency_dict = analysis_dict[input_token]
    tokens = list(frequency_dict.keys())
    weights = list(frequency_dict.values())
    return choices(tokens, weights=weights, k=1)[0]
```

```python
seed = "Cowboy"
output_len = 100
for i in range(output_len - len(seed)):
    chunk = seed[-source_token_size:]
    seed += predict_next_token(chunk)

seed
```
